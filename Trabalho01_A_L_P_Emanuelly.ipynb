***QUESTÃO 4***
*   **Raciocínio**

1. Ler um número inteiro.
2. Verificar o resto da divisão do número por 2.
3. Se o resto for igual a zero, o número vai ser par.
4. Caso contrário, o número é ímpar.


*   **Desafio de raciocínio:**
 qual operador permite descobrir se uma divisão possui resto?

> O operador %

***QUESTÃO 5***
*   **Raciocínio**

1. Ler duas notas.
2. Calcular a média somando as duas notas e dividindo por 2.
3. Se a média for maior ou igual a 7, definir como "Aprovado".
4. Se não for, mas a média for maior ou igual a 5, definir como "Recuperação".
5. Caso contrário (menor que 5), definir como "Reprovado".
6. Mostrar a média e a situação final.

````python
nota1 = float(input("Digite a primeira nota: "))
nota2 = float(input("Digite a segunda nota: "))
media = (nota1 + nota2) / 2
if media >= 7:
    situacao = "Aprovado"
elif media >= 5:
    situacao = "Recuperação"
else:
    situacao = "Reprovado"
print(f"Média: {media:.1f}")
print(f"Situação: {situacao}")
````

4. Testar se esse número digitado é par. Se for, somar 1 à variável ```pares```.

***QUESTÃO 14***
*   **Raciocínio**

1. Definir o número secreto fixo com 7.
2. Criar uma variável contadora chamada ```tentativas``` começando com 0.
3. Iniciar um laço infinito.
4. Dentro do laço, ler o palpite do usuário e somar 1 no contador de tentativas.
5. Se o palpite for igual a 7, mostrar mensagem de parabéns e quebrar o laço.
6. Se o palpite for menor que 7, dar a dica para tentar um número maior.
7. Se o palpite for maior que 7, dar a dica para tentar um número menor.
8. Após o acerto, mostrar o total de tentativas necessárias.

````python
numero_secreto = 7
tentativas = 0
while True:
    palpite = int(input("Tente descobrir o número:"))
    tentativas += 1
    if palpite == numero_secreto:
        print("Parabéns, você acertou.")
        break
    elif palpite < numero_secreto:
        print("Tente um número maior.")
    else:
        print("Tente um número menor")
    print(f"Além disso, foram necessárias{tentativas} tentativas.")
````

***QUESTÃO 15***
*   **Raciocínio**

1. Criar 4 variáveis de controle: ```total_numeros```, ```positivos```,  ```negativos``` e  ```soma```.
2. Abrir um laço de repetição contínua.
3. Ler um número inteiro de dentro do laço.
4. Verificar se o número é igual a 0. Se for, interromper o laço imediatamente.
5. Se não for zero, incrementar o ```total_numeros``` em 1 e somar o valor digitado na variável  ```soma```.
6. Testar se o número é maior que 0: se sim, somar 1 na variável ```positivos```; se não, somar 1 na variável ```negativos```.  
7. Exibir todos os relatórios formatados após o término.

````python
total = 0
positivos = 0
negativos = 0
soma = 0
while True:
   numero = int(input("Digite um número:"))
   if numero == 0:

       break
   total += 1
   soma += numero
   if numero > 0:
       positivos += 1
   else:
       negativos += 1
print(f"Quantidade:{total}")
print(f"Positivos:{positivos}")
print(f"Negativos:{negativos}")
print(f"Soma:{soma}")
````
